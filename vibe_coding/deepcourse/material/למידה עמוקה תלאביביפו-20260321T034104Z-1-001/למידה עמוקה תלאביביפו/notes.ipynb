* [הרצאה](https://www.youtube.com/watch?v=EPJI8-VN5kk) שבתחילתה הרבה דוגמאות למה שלמידה עמוקה עושה היום בחיים שלנו

```python
from IPython.display import IFrame

def youtube(url, width=600, height=600):
  i1 = url.index('v=') + 2
  try:
    i2 = url.index('&', i1)
  except ValueError:
    i2 = len(url)
  id = url[i1:i2]

  display(IFrame('https://www.youtube.com/embed/' + id,
          width=width,
          height=height))
```
