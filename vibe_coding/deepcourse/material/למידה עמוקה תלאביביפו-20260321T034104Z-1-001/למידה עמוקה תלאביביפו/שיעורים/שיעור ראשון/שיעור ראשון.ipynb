#### למידה מונחית - Supervised learning

<div dir=rtl>

**למידה מונחית** היא למידה, שבה הקלט לאלגוריתם הלומד, הוא אוסף של זוגות: דוגמה, והפתרון הנכון לה.

לדוגמה: בעיה שבה אנחנו רוצים לזהות ספרות הכתובות בכתב יד. הבעיה נקראת OCR - Optical Character Recognition.
במקרה זה, האלגוריתם יקבל אוסף של זוגות - תמונה, והספרה שכתובה בה:


```python
# @title Code to show some handwritten digits images, and their lables
import tensorflow as tf
import matplotlib.pyplot as plt

mnist = tf.keras.datasets.mnist

(train_x, train_y), _ = mnist.load_data()

N = 1 + 7

plt.figure(figsize=(2, 4))

plt.subplot(N, 2, 1)
plt.axis('off')
plt.text(0.5, 0.5, 'X', ha='center', va='center')
plt.subplot(N, 2, 2)
plt.axis('off')
plt.text(0.5, 0.5, 'Y', ha='center', va='center')

for i in range(1, N):
  plt.subplot(N, 2, 2*i+1)
  plt.imshow(train_x[i], cmap='binary')
  plt.axis('off')

  plt.subplot(N, 2, 2*i+2)
  plt.text(0.5, 0.5, f'{train_y[i]}', ha='center', va='center')
  plt.axis('off')
```

<div dir=rtl>

האתגר של האלגוריתם הלומד, הוא למצוא פונקציה, חוק התאמה, בין תמונה- X לתיוג הנכון שלה -Y.

הלמידה נקראית מונחית, או מפוקחת, בגלל שאנחנו מספקים את התשובה הנכונה לכל דוגמה, כלומר, מנחים את האלגוריתם, איך הוא אמור לפעול (ככה לימדו אותי לצבוע עצים בעבודות רס"ר בצבא - אתה רואה את העץ הזה, ככה צובעים אותו. ואת ההעץ הזה? ככה. תמשיך ככה..)

## בעית למידה מונחית, כבעיית קירוב פונקציה

<div dir=rtl>

נחזור עכשיו לדוגמת זיהוי הספרות, מתמונות של כתב יד. ה dataset שנקרא [mnist](https://en.wikipedia.org/wiki/MNIST_database) הוא אוסף תמונות מתויגות של כתב יד, שמשמש כמקרה בוחן פשוט, ללמידה מונחית.

הנה מספר נתונים על המאגר:

```python
# @title Code for MNIST information
import tensorflow as tf
import matplotlib.pyplot as plt

mnist = tf.keras.datasets.mnist

(train_x, train_y), (test_x, test_y) = mnist.load_data()

print(f'Number of labeled training examples: {len(train_x):,}')
print(f'Number of labeled testing examples: {len(test_x):,}')
shape = train_x[0].shape
print(f'Dimention of each handwriten digit image: {shape}')
print(f'Or as a one dimentional vector of {shape[0] * shape[1]} elements.')

idx = 6 # change this number to see others
plt.imshow(test_x[idx], cmap='binary')
plt.title(f'An example of a character image ({test_y[idx]})')
plt.show()
```

<div dir=rtl>

האתגר הוא לכתוב קוד, שמסתכל על הרבה דוגמאות מתויגות, זוגות של תמונה והסיפרה האמיתית שכתובה שם, וללמוד את החוקיות. כלומר, ללמוד מיפוי מתמונה לתיוג. במילים אחרות, הפלט של קוד כזה, אמור להיות פונקציה שה X שלה הוא תמונה, וה - Y שלה הוא תיוג. בצורה פורמלית:

$$
f:\{0,1,2,\dots,255\}^{28 \times 28} \rightarrow \{0,1,2, \dots,9\}
$$

כלומר, פונקציה שממפה מטריצות בגודל 28 על 28 שהערכים שלהם הם מספרים בין 0 ל 255, לסיפרה.

אם נעשה רלקסציה, ונחשוב על המטריצות כמטריצות כלליות של מספרים ממשיים, אפשר לראות את הפונקציה כך:

$$
f:\mathbb{R}^{28 \times 28} \rightarrow \{0,1,2, \dots,9\}
$$

ואם נתעלם מהעובדה שמדובר בתמונה או מטריצה, ונשטח את כל הערכים של התמונה, לווקטור חד ממדי, נקבל:
$$
f:\mathbb{R}^{784} \rightarrow \{0,1,2, \dots,9\}
$$

<div dir=rtl>

אנחנו רגילים לעסוק בפונקציות ממספר למספר, או מזוג מספרים למספר. פונקציות כאלה אחנו יכולים לתאר בצורה ציורית, במערכות צירים.

הפונקציה במקרה שלנו, מקבלת ווקטור בממד 784 ומחזירה אבר מקבוצה של עשרה אברים.

הוויזואלזציות שלנו מוגבלות לשלושה ממדים, בגלל שזה העולם בו אנו חיים.

אבל אלגברה וקלקולוס, לא מוגבלים לשלושה ממדים. שם, אנחנו יכולים לעשות חישובים בעולמות עם כמה ממדים שאנחנו רוצים.

אבל כעת, בשביל לקבל אינטואיציה על מה שאנחנו עושים פה, נרד לבעיה הרבה יותר פשוטה, אותה נוכל לתאר עם מעט ממדים ואותה גם נוכל לצייר.

נניח שבמקום זוגות של תמונה ותיוג, היינו מקבלים סתם זוגות של מספרים. X הוא מספר, ו Y הוא מספר.

ושוב, כמו בכל בעיית למידה מונחית, המטרה שלנו היא ללמוד את חוק ההתאמה בינהם, כך שכאשר נראה X חדש, כזה שלא ראינו בזמן האימון, נדע איזה Y להתאים לו.

### מציאת הפרבולה הטובה ביותר

<div dir=rtl>

במקרה המפושט שנדון בו עכשיו, הקלט לאלגוריתם הוא אוסף דוגמאות שכל דוגמה היא זוג סדור של מספרים, X ו Y.

$$
\text{training set} = \{(x_1, y_1), (x_2, y_2), ..., (x_n, y_n) \}
$$

$x_i$ ו $y_i$ כאן, הם פשוט מספרים שנשארים קבועים במשך כל תהליך הלימוד. לדוגמה, קבוצת דוגמאות ספציפית יכולה להראות כך:
$$
\text{training set} = \{(17, 6), (-3, 0.6), (170, 600), (0,0)\}
$$

<div dir=rtl>

אנחנו ננסה למצוא פונקציה מקרבת. כלומר פונקציה שבנקודות האלה, תהיה קרובה.

ראשית עלינו להחליט על קבוצה של פונקציות, שמתוכן נבחר את הפונקציה המקרבת שלנו.

בשביל הפשטות, בואו נניח כרגע שאנחנו עוסקים רק בפרבולות מהצורה:

$$
f(x) = ax^2 + bx - 2
$$

זאת משפחה של פרבולות. כל פרבולה מהמשפחה הזו, מוגדרת בצורה חד ערכית, ע"י ערכי ה a וה b  שלה.

הנה [הדגמה אינטראקטיבית](https://www.geogebra.org/calculator/ae8f5shc)  של האתגר העומד מולנו.

<div dir=rtl>

כעת, אנו זקוקים למדד, שיתאר את איכות הקירוב של הפונקציה לדוגמאות. למדד הזה קוראים loss או cost, ואפשר להגדיר אותו בהרבה אופנים. אנחנו נרצה שיתאר סוג של מרחק מהפונקציה לדוגמאות. כשכל הדוגמאות יהיו על הפונקציה, נרצה שהמרחק הזה יהיה 0 וככל שהם יהיו יותר רחוקות ממנה, המדד יהיה מספר חיובי יותר גדול. סוג של מטריקה.

אפשרות אחת היא להגדיר את המדד הזה, פשוט כסכום המרחקים האנכיים:
$$ cost(a,b) = \sum_{i=1}^n{ \left|f_{a,b}(x_i)-y_i \right|}$$
שימוש בערך מוחלט יכול להיות בעייתי בכל מני הקשרים. לדוגמה, זו לא פונקציה גזירה, וכפי שנראה בהמשך, אנחנו נרצה להשתמש בכלים של קלקולוס, כי מדובר בבעיית אופטימיזציה. לכן, בדרך כלל זו לא תהיה ההגדרה שבה נבחר להגדיר את המדד שלנו.

שימו לב שכשכתבנו $f_{a,b}(x)$, הכוונה היתה לאותה פונקציה שמוגדרת ע"י בחירת פרמטרים a ו b ספציפיים. במובן מסויים יש פה שינוי תפקיד, כי בשביל cost, a ו b משמשים משתנים, ובשביל f הם משמשים פרמטרים.

כידוע, אין הבדל מתמטי בין פרמטר למשתנה, ההבדל הוא רק על מי אנחנו חושבים כקבוע בהקשר מסויים, ולמי אנחנו מתייחסים כערך חופשי, או ערך רץ.


הגדרה יותר פופולרית לפונקציה המחיר שלנו, תהיה כזו:
$$ cost(a,b) = \sum_{i=1}^n{ \left(f_{a,b}(x_i)-y_i \right)^2}$$

במדד זה נותנים יותר חשיבות למרחקים גדולים. לדוגמה, מרחק גדול פי שניים, ישפיע פי 4 במדד.

ההגדרה הזו, בניגוד לקודמת, היא גזירה ונח לעבוד איתה.

אפשר גם לחלק במספר הנקודות ולדבר בעצם על ממוצע ריבועי המרחקים (הגדרה דומה לשונות בהסתברות) באופן הבא:

$$
cost(a,b) = \frac{1}{n}\sum_{i=1}^n{ \left(f_{a,b}(x_i)-y_i \right)^2}
$$

אבל זה לא משנה את האופטימיזציה בגלל שזו הכפלת פונקציה במספר קבוע (למעט שיקולים של חישובים נומריים)

### פונקציית ה cost והמשמעות שלה

<div dir=rtl>
הפונקציה cost מקבלת זוג ערכים, עבור a ו b, ומחזירה מספר אי שלילי. שמייצג את המרחק של הפונקציה המוגדרת ע"י a  ו b מאוסף ספציפי של דוגמאות.

cost: \mathbb{R}^2 → \mathbb{R}_{\geq 0}
$$

$$
cost(a,b) = distance \left[f_{a.b}(x), \{(x_1, y_1), ... ,(x_n, y_n)\} \right]

### אלגוריתם אופטימיזציה פשוט - gradient decent

<div dir=rtl>

המטרה שלנו היא, אם כן, למצוא את המינימום של פונקצית המחיר. במקרה שלנו זו פונקציה שמקבלת שני ארגומנטים, אבל במציאות הפונקציה הזאת תקבל, כנראה, הרבה יותר. אנחנו זקוקים לאלגוריתם שידע למצוא מינימום של פונקציה רבת משתנים.

אחד האלגוריתמים הידועים והפשוטים נקרא gradient decent. האלגוריתם, במילים, הולך ככה:

1. בחר נקודה להתחיל ממנה
1. מצא את הכיוון הכי תלול לרדת בו
1. לך צעד בכיוון הזה
1. חזור ל2 עד שאתה מחליט שמספיק.

<div dir=rtl>

האלגוריתם, כפי שתואר, מותיר מספר שאלות:
1. איך מוצאים את הכיוון הזה, שבו הכי תלול לרדת?
2. מה זה מספיק? איך נדע מתי לעצור?


<div dir=rtl>

נתחיל מלענות על השאלה הראשונה: [גרדיאנט](https://he.wikipedia.org/wiki/%D7%92%D7%A8%D7%93%D7%99%D7%90%D7%A0%D7%98).

כפי שזכור לכם (?) מלימודי הקלקולוס שלכם, הגרדיאנט הוא ווקטור הנגזרות החלקיות ויש לו את התכונה להצביע תמיד על הכיוון התלול ביותר.

$$
f: \mathbb{R}^n → \mathbb{R}
$$

$$
\vec{a_0} \in \mathbb{R}^n
$$

$$
\nabla f(\vec{a_0}) \equiv\left(\frac{\partial f}{\partial x_1}(\vec{a_0}), \frac{\partial f}{\partial x_2}(\vec{a_0}), \ldots, \frac{\partial f}{\partial x_n}(\vec{a_0})\right)
$$

<div dir=rtl>


כשזזים קצת מנקודה מסויימת, השינוי בגובה של הפונקציה, יהיה המכפלה הסקלרית של ווקטור התזוזה, בגרדיאנט:

$$
\Delta f =   \left< \nabla f(\vec{a_0}), \Delta \vec{x} \right>  
$$

<div dir=rtl>

מבחינה גאומטרית, מכפלה סקלרית היא מכפלת אורך ווקטור באורך של ההיטל של הווקטור השני עליו:
$$
\left < \vec{x}, \vec{y}   \right> = ||\vec{x}|| \cdot ||\vec{y}|| cos (\alpha)
$$

כאשר $\alpha$ היא הזווית בין הווקטורים.
עבור ווקטור תזוזה באורך קבוע, המכפלה הסקלארית תהיה הגדולה ביותר, כאשר $\nabla f(a_0)$ ו $\Delta \vec{x}$ יהיו באותו הכיוון, כלומר כאשר  $cos(α) = 1$ ואז $α=0$.



כאשר הכיוון הוא מאונך לגרדיאנט, המכפלה היא 0, וזה הכיוון של קו הגובה (עקומת אדישות, קונטור) וכאשר הכיוון של התזוזה הוא נגדי לגרדיאנט, זה מה שמפחית את ערך הפונקציה הכי הרבה, וזה הכיוון שאנחנו מחפשים לאופטימיזציית מינימום.



<div dir=rtl>

```python
# @title show_image() code
from IPython.display import HTML

def show_image(url, width=400):
  i = url.find('file')
  j = url.find('/', i+7)
  url = url[:i] + 'uc?id=' + url[i+7: j]
  html = HTML(f'<center><img src="{url}" width={width} /></center>')
  display(html)

show_image('https://drive.google.com/file/d/1_pmaYLISwSowNgiCH2N3XvdGg_thXpSX/view?usp=drive_link', 150)
```

```python
# @title youtube() code
from IPython.display import IFrame

def youtube(url, width=600, height=600):
  i1 = url.find('v=')+2
  i2 = url.find('&', i1)
  if i2 == -1:
    i2 = len(url)

  vid = url[i1 : i2]
  url = 'https://www.youtube.com/embed/' + vid

  ifr = IFrame(url, width, height)
  display(ifr)

youtube('https://www.youtube.com/watch?v=WXuK6gekU1Y&t=7s', 150, 100)
```
