## מבוא
בעשור האחרון, למידה עמוקה פרצה לחיינו בסערה. הציבור הרחב מקבל המון טכנולוגיה חדשה ורובו לא מודע לעובדה, שכולה ממקור אחד - שימוש ברשתות נוירונים לומדות.
נהיגה אוטונומית, זיהוי דיבור, תירגום, יצירת תמונות, שיחה כללית, ייצור קוד, הבנת תמונות, ייצור סרטים ועוד ועוד ועוד.

בקורס הזה נלמד קצת את ההיסטוריה של התחום, את הרעיון המרכזי של למידה מונחית כקרוב של פונקציה מתוך משפחה של פונקציות. אלגוריתם אופטימיזציה כמעט טריוויאלי בשם gradient decent.
אחרי מבוא זה, נלמד את הנושאים הבאים.



In the past decade, deep learning has stormed into our lives like a whirlwind. The general public is embracing a plethora of new technologies, often unaware that they all stem from one source - the use of neural networks for learning. Autonomous driving, speech recognition, translation, image generation, natural language processing, code generation, image understanding, film production, and much more, all owe their advancements to the utilization of neural networks.

In this course, we will delve into the history of the field and explore the central concept of supervised learning as an aproximation of a function by a candidate from a function set. We will introduce an almost trivial optimization algorithm called gradient descent. Following this introduction, we will cover the following topics.
